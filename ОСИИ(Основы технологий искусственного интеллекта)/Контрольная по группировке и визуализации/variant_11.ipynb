**20б. Ловкость — не менее пяти неравных групп; среднее ускорение по группам ловкости**

```python
data['agility_group'] = pd.cut(data['Agility'], bins=[0, 30, 45, 60, 75, 100])
res1 = data.groupby('agility_group', observed=False)['Acceleration'].mean()
print(res1)
res1.plot(kind='bar')
plt.title('Среднее ускорение по группам ловкости')
plt.show()
```

**20в. Короткие передачи — шесть равных групп; средняя ценность по группам коротких передач**

```python
data['short_group'] = pd.cut(data['ShortPassing'], bins=6)
res2 = data.groupby('short_group', observed=False)['Value'].mean()
print(res2)
res2.plot(kind='bar')
plt.title('Средняя ценность по группам коротких передач')
plt.show()
```
